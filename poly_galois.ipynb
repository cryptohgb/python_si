```python
def random_irreducible_poly( field, degree ):
    coeffs = [field(1)] + [ choice( list(field.elements) ) for _ in range(degree) ]
    m = galois.Poly( coeffs, field=field )
    while len(m.factors()[0])>1:
        coeffs = [field(1)] + [ choice( field.elements ) for _ in range(degree) ]
        m = galois.Poly( coeffs, field=field )
    return m
```

```python
t = 10

g = random_irreducible_poly( ff, t )
g
```
