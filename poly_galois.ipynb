```python
from random import choice
```

```python
def random_irreducible_poly( field, degree ):
    coeffs = [field(1)] + [ choice( list(field.elements) ) for _ in range(degree) ]
    m = galois.Poly( coeffs, field=field )
    while len(m.factors()[0])>1:
        coeffs = [field(1)] + [ choice( field.elements ) for _ in range(degree) ]
        m = galois.Poly( coeffs, field=field )
    return m
```

```python
m = 8   # GF(2**8)
n = 40  # Stützstellen
t = 3   # Grad von irred. g
# beachte n >= m*t
```

```python
g = random_irreducible_poly( ff, t )
g
```

```python
a = [ ff(i) for i in range( 2, n+2 ) ]
```

```python
g_of_a = g(a)
g_of_a
```

```python
vandermont = []
for j in range(t):
    vandermont = vandermont + [[ g_of_a[i]**j for i in range(n)]]

vandermont = ff(vandermont)
vandermont
```

```python
diag = np.diag(ff([ ff(1)/(g_of_a)[i] for i in range(n) ]))
diag = ff(diag)
diag
```

```python
coeffs = list(g.coefficients())
left_matrix = []
for j in range(t):
    left_matrix = left_matrix + [ j*[ff(0)]+coeffs[:-1-j] ]
left = ff(left_matrix).transpose()
```

```python
parity_check_H = left @ vandermont @ diag
parity_check_H
```

```python
def int_2_m_bit( n, m ):
    bits = []
    for _ in range(m):
        bits = [n%2]+bits
        n = n//2
    return bits
```

```python
int_2_m_bit( 15, 8 )
```

```python
parity_check_H_bin_unstacked = [ np.array( [ int_2_m_bit( int(x), m ) for x in parity_check_H[j] ] ).transpose() for j in range(t) ]
```

```python
parity_check_H_bin = np.concatenate( parity_check_H_bin_unstacked, axis=0 )
parity_check_H_bin.shape
```
